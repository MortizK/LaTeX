# Aufgabe 5.4: Wahrscheinlichkeitsfunktion und Erwartungstreue

In einem gut gefüllten Saal befinden sich N Personen, die mit 1 bis N durchnummeriert sind. Herr
Maier von der Gewerbeaufsicht kennt N nicht und wird vom Veranstalter mit allerlei Tricks am Betreten
des Saals gehindert. Herr Maier, der N ermitteln möchte, ist lediglich in der Lage, die Nummer eines
Teilnehmers, der zufällig den Saal verlässt, zu erkennen. Fassen Sie diese Nummer als Realisation x
einer Zufallsvariablen X auf.

```python
import numpy as np

N = 300

def schaetzung(x):
    return 2*x - 1

def expectation():
    schaetzungen = []
    mean_schaetzung = []
    people = np.arange(0, N)
    for i in range(N):
        # x randomly chosen from people and remove that person
        x = np.random.choice(people)
        people = np.delete(people, np.where(people == x))
        schaetzungen.append(schaetzung(x))
        mean_schaetzung.append(np.mean(schaetzungen))
    return mean_schaetzung

import matplotlib.pyplot as plt
for i in range(3):
    mean_schaetzung = expectation()
    plt.plot(mean_schaetzung, label=f'Mittelwert Lauf {i+1}')
plt.axhline(y=N, color='r', linestyle='--', label='Wahrer Wert')
plt.xlabel('Anzahl der Schaetzungen')
plt.ylabel('Schaetzungswert')
plt.title('Schaetzungen und deren Mittelwert')
plt.legend()
plt.show()
```
