## Lösungsmenge für Volumen und Gewicht

```python
# Volumenlimit 2500

def testlimit(a,b,c,d):
    return (a + b + c + d < 4)
    
    
csp.addConstraint(testlimit, ["A","B","C","D"])
    
# Gewichtslimit 2800

def testlimit_42(a,b,c,d):
    return (a + b + c + d < 42)
    
    
csp.addConstraint(testlimit_42, ["A","B","C","D"])
    
```

```python
csp.getSolutions()
```
