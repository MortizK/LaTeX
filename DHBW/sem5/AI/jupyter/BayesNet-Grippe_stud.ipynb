# 1. Netzstruktur verstehen

Im Rahmen eines medizinischen Frühwarnsystems soll mit Hilfe eines Bayes’schen Netzes modelliert werden, wie wahrscheinlich es ist, dass eine Person an Grippe erkrankt ist. Die Prognose soll auf Basis der beobachteten Symptome (Fieber und/oder Husten?) sowie dem Verhalten (sucht die Person einen Arzt auf?) erfolgen.

Das Ziel ist es, das Netz zu analysieren und mit Hilfe von bedingten Wahrscheinlichkeiten diagnostische und prädiktive Inferenz durchzuführen.

```text
Variable    Bedeutung	          Zustände (Werte)
#------------------------------------------------#
G(rippe)    Person hat Grippe     0 = nein, 1 = ja
F(ieber)    Person hat Fieber     0 = nein, 1 = ja
H(usten)    Person hat Husten     0 = nein, 1 = ja
A(rzt)      Person geht zum Arzt  0 = nein, 1 = ja
```

Die Abhängigkeiten sind wie folgt definiert:
* Grippe beeinflusst Fieber und Husten
* Fieber und Husten beeinflussen gemeinsam die Wahrscheinlichkeit für BesuchArzt

Folgende Wahrscheinlichkeiten sind gegeben:

Prior:
------
```text
* p(G=1) = 0.10
* p(G=0) = 0.90
```

Bedingte Wahrscheinlichkeiten:
------------------------------

* Fieber | Grippe:

```text
Grippe	p(F=1)
#-------------------#
1       0.80
0       0.10
```

* Husten | Grippe:

```text
Grippe	p(H=1)
#-------------------#
1       0.70
0       0.20
```

* Arzt | Fieber, Husten:

```text
Fieber	Husten	p(A=1 | F, H)
#---------------------------#
1       1       0.90
1       0       0.75
0       1       0.60
0       0       0.10
```
